```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import numpy as np

# ===============================================================
# LOAD FUSED DATASET (train / val / test)
# ===============================================================

df = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/"
    "MODEL_SOCAT_AND_FLOATS_v2/processing_steps/pco2_train_data.csv"
)

print(df["source_flag"].value_counts())

# ===============================================================
# SOURCE TYPES
# ===============================================================

source_labels = {
    0: "SOCAT",
    1: "FLOATS",
    2: "FUSED"
}

source_flags = sorted(df["source_flag"].unique())

# ===============================================================
# SET UP MAP
# ===============================================================

fig = plt.figure(figsize=(7, 7), dpi=300)
ax = fig.add_subplot(1, 1, 1, projection=ccrs.SouthPolarStereo())

ax.set_extent([-180, 180, -90, -30], crs=ccrs.PlateCarree())

# Base features
ax.coastlines(resolution='110m', linewidth=0.6)
ax.add_feature(cfeature.LAND, facecolor='lightgray', zorder=0)
ax.gridlines(draw_labels=False, linewidth=0.2, color='gray', alpha=0.5)

# ===============================================================
# COLORS & MARKERS (MATCHED TO 3 DATA TYPES)
# ===============================================================

colors = {
    0: "#1f77b4",   # SOCAT → blue
    1: "#ff7f0e",   # FLOATS → orange
    2: "#2ca02c"    # FUSED → green
}

markers = {
    0: "o",   # SOCAT → circles
    1: "^",   # FLOATS → triangles
    2: "s"    # FUSED → squares
}

# ===============================================================
# PLOT EACH SOURCE TYPE
# ===============================================================

for flag in source_flags:

    df_src = df[df["source_flag"] == flag]

    ax.scatter(
        df_src["longitude"],
        df_src["latitude"],
        s=8,
        label=source_labels[flag],
        color=colors[flag],
        marker=markers[flag],
        transform=ccrs.PlateCarree(),
        alpha=0.8,
        zorder=2
    )

# ===============================================================
# LEGEND
# ===============================================================

ax.legend(
    title="Pixel Source Type",
    loc="lower left",
    fontsize=7,
    title_fontsize=8,
    frameon=True
)

# ===============================================================
# SAVE & SHOW
# ===============================================================

plt.tight_layout()
plt.savefig(
    "figs/socat_floats_fused_pixels_map.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
