```python
# ============================================================
# === COMBINED SOCAT + FLOATS MC FLUX WITH UNCERTAINTY ========
# === Identical methodology, single unified analysis cell ===
# ============================================================

import pandas as pd
import numpy as np
import pyseaflux

# ============================================================
# 1) LOAD DATA (SOCAT + FLOATS)
# ============================================================

# ---------- SOCAT ----------
predictions_socat = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/pco2_prediction_matrix_SOCAT_only_MCensemble.csv"
)

predictions_socat["month_center"] = pd.to_datetime(predictions_socat["month_center"])
predictions_socat = predictions_socat[
    ~((predictions_socat["month_center"].dt.year == 2023) &
      (predictions_socat["month_center"].dt.month == 2))
]

socat_df = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/MODEL_SOCAT_ONLY_OBS_v2/processing_steps/SOCATv2025_SO_feature_engineered_monthly_1deg_pCO2.csv",
    low_memory=False
)

# ---------- FLOATS ----------
predictions_floats = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/pco2_prediction_matrix_FLOATS_only_MCensemble.csv"
)

predictions_floats["month_center"] = pd.to_datetime(predictions_floats["month_center"])
predictions_floats = predictions_floats[
    ~((predictions_floats["month_center"].dt.year == 2023) &
      (predictions_floats["month_center"].dt.month == 2))
]

floats_df = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/MODEL_FLOATS_v2/processing_steps/BGC-Argo_SO_feature_engineered_monthly_1deg.csv",
    low_memory=False
)

floats_df.rename(columns={"pco2": "FLOATS_pco2_median"}, inplace=True)

# ============================================================
# 2) OBS UNCERTAINTY
# ============================================================

SOCAT_SIGMA = 5.0
socat_df_low  = socat_df.copy()
socat_df_high = socat_df.copy()
socat_df_low["pco2"]  -= SOCAT_SIGMA
socat_df_high["pco2"] += SOCAT_SIGMA

floats_df_low  = floats_df.copy()
floats_df_high = floats_df.copy()
floats_df_low["FLOATS_pco2_median"]  -= floats_df_low["FLOATS_pco2_error_median"]
floats_df_high["FLOATS_pco2_median"] += floats_df_high["FLOATS_pco2_error_median"]

# ============================================================
# 3) MODEL MC ENSEMBLE UNCERTAINTY
# ============================================================

pred_socat_low  = predictions_socat.copy()
pred_socat_high = predictions_socat.copy()
sigma_model_socat = predictions_socat["ensemble_pco2_std"].values
pred_socat_low["ensemble_pco2_pred"]  -= sigma_model_socat
pred_socat_high["ensemble_pco2_pred"] += sigma_model_socat

pred_floats_low  = predictions_floats.copy()
pred_floats_high = predictions_floats.copy()
sigma_model_floats = predictions_floats["ensemble_pco2_std"].values
pred_floats_low["ensemble_pco2_pred"]  -= sigma_model_floats
pred_floats_high["ensemble_pco2_pred"] += sigma_model_floats

# ============================================================
# 4) FLUX FUNCTIONS
# ============================================================

def flux_from_pco2(df, temp, sal, pco2_sw):
    wind = df["wind_speed"].values
    air  = df["atm_co2"].values
    pres = 1013.25
    k_w = pyseaflux.gas_transfer_velocity.k_Ho06(wind, temp)
    flux_mol = pyseaflux.flux_calculations.flux_bulk(
        temp, sal, pco2_sw, air, pres, k_w
    )
    return flux_mol * 12.01

def flux_socat_obs(df):
    return flux_from_pco2(df, df["SOCAT_temperature"].values,
                           df["SOCAT_salinity"].values, df["pco2"].values)

def flux_socat_pred(df):
    return flux_from_pco2(df, df["SOCAT_temperature"].values,
                           df["SOCAT_salinity"].values, df["ensemble_pco2_pred"].values)

def flux_floats_obs(df):
    return flux_from_pco2(df, df["FLOATS_temperature_median"].values,
                           df["FLOATS_salinity_median"].values, df["FLOATS_pco2_median"].values)

def flux_floats_pred(df):
    return flux_from_pco2(df, df["FLOATS_temperature_median"].values,
                           df["FLOATS_salinity_median"].values, df["ensemble_pco2_pred"].values)

# ============================================================
# 5) GRID + AREA WEIGHTING (SHARED)
# ============================================================

def bin_flux(df, lat_col, lon_col, time_col, flux_col):
    df = df.copy()
    df["lat_bin"] = df[lat_col].astype(int)
    df["lon_bin"] = df[lon_col].astype(int)
    df["year"]  = df[time_col].dt.year
    df["month"] = df[time_col].dt.month
    return df.groupby(["lat_bin","lon_bin","year","month"])[flux_col].mean().reset_index(name="mean_flux")

R = 6371000
deg2rad = np.pi/180
lat_centers = np.arange(-89.5, 90.5, 1)[:-1]

band_areas = 2*np.pi*R**2 * (np.sin((lat_centers+0.5)*deg2rad)-np.sin((lat_centers-0.5)*deg2rad))
cell_area_by_lat = pd.Series(band_areas/360, index=np.arange(-89,90,1))

def area_weighted_monthly_mean(binned):
    df = binned.merge(cell_area_by_lat.rename("cell_area_m2"),
                      left_on="lat_bin", right_index=True, how="left")
    df["weighted"] = df["mean_flux"]*df["cell_area_m2"]
    out = df.groupby(["year","month"],as_index=False).agg(
        total_flux=("weighted","sum"), total_area=("cell_area_m2","sum")
    )
    out["aw_mean_flux"] = out["total_flux"]/out["total_area"]
    out["date"] = pd.to_datetime(dict(year=out["year"],month=out["month"],day=15))
    return out[["year","month","date","aw_mean_flux"]]

# ============================================================
# 6) PROCESS SOCAT
# ============================================================

for df in [socat_df, socat_df_low, socat_df_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_socat_obs(df) * df["time"].dt.days_in_month

socat       = area_weighted_monthly_mean(bin_flux(socat_df,"lat_center","lon_center","time","flux_month"))
socat_low   = area_weighted_monthly_mean(bin_flux(socat_df_low,"lat_center","lon_center","time","flux_month"))
socat_high  = area_weighted_monthly_mean(bin_flux(socat_df_high,"lat_center","lon_center","time","flux_month"))

socat.columns      = ["year","month","date","SOCAT"]
socat_low.columns  = ["year","month","date","SOCAT_low"]
socat_high.columns = ["year","month","date","SOCAT_high"]

for df in [predictions_socat, pred_socat_low, pred_socat_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_socat_pred(df) * df["time"].dt.days_in_month

pred_socat       = area_weighted_monthly_mean(bin_flux(predictions_socat,"latitude","longitude","time","flux_month"))
pred_socat_low   = area_weighted_monthly_mean(bin_flux(pred_socat_low,"latitude","longitude","time","flux_month"))
pred_socat_high  = area_weighted_monthly_mean(bin_flux(pred_socat_high,"latitude","longitude","time","flux_month"))

pred_socat.columns      = ["year","month","date","SOCAT_predictions"]
pred_socat_low.columns  = ["year","month","date","SOCAT_predictions_low"]
pred_socat_high.columns = ["year","month","date","SOCAT_predictions_high"]

# ============================================================
# 7) PROCESS FLOATS
# ============================================================

for df in [floats_df, floats_df_low, floats_df_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_floats_obs(df) * df["time"].dt.days_in_month

floats       = area_weighted_monthly_mean(bin_flux(floats_df,"lat_center","lon_center","time","flux_month"))
floats_low   = area_weighted_monthly_mean(bin_flux(floats_df_low,"lat_center","lon_center","time","flux_month"))
floats_high  = area_weighted_monthly_mean(bin_flux(floats_df_high,"lat_center","lon_center","time","flux_month"))

floats.columns      = ["year","month","date","floats"]
floats_low.columns  = ["year","month","date","floats_low"]
floats_high.columns = ["year","month","date","floats_high"]

for df in [predictions_floats, pred_floats_low, pred_floats_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_floats_pred(df) * df["time"].dt.days_in_month

pred_floats       = area_weighted_monthly_mean(bin_flux(predictions_floats,"lat_center","lon_center","time","flux_month"))
pred_floats_low   = area_weighted_monthly_mean(bin_flux(pred_floats_low,"lat_center","lon_center","time","flux_month"))
pred_floats_high  = area_weighted_monthly_mean(bin_flux(pred_floats_high,"lat_center","lon_center","time","flux_month"))

pred_floats.columns      = ["year","month","date","floats_predictions"]
pred_floats_low.columns  = ["year","month","date","floats_predictions_low"]
pred_floats_high.columns = ["year","month","date","floats_predictions_high"]

# ============================================================
# 8) MERGE EVERYTHING
# ============================================================

monthly_flux_summary = (
    socat
    .merge(socat_low,on=["year","month","date"])
    .merge(socat_high,on=["year","month","date"])
    .merge(pred_socat,on=["year","month","date"])
    .merge(pred_socat_low,on=["year","month","date"])
    .merge(pred_socat_high,on=["year","month","date"])
    .merge(floats,on=["year","month","date"])
    .merge(floats_low,on=["year","month","date"])
    .merge(floats_high,on=["year","month","date"])
    .merge(pred_floats,on=["year","month","date"])
    .merge(pred_floats_low,on=["year","month","date"])
    .merge(pred_floats_high,on=["year","month","date"])
)

# ============================================================
# 9) SCALE TO Pg C yr⁻¹
# ============================================================

#OCEAN_AREA_M2 = 7e13 // commented out because we now compute it cleanly using an ocean mask

AREA_CSV = "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/ocean_mask/ocean_area_south_of_30S.csv"

print("Loading ocean area south of 30°S...")
OCEAN_AREA_M2 = pd.read_csv(AREA_CSV)["ocean_area_m2"].iloc[0]
print(f"   Using OCEAN_AREA_M2 = {OCEAN_AREA_M2:.3e} m²")

for col in monthly_flux_summary.columns:
    if col not in ["year","month","date"]:
        monthly_flux_summary[col] = (
            monthly_flux_summary[col] * 12 * OCEAN_AREA_M2 / 1e15
        )

# ============================================================
# 10) SIGMA
# ============================================================

monthly_flux_summary["SOCAT_sigma_PgC_yr"] = 0.5*(monthly_flux_summary["SOCAT_high"]-monthly_flux_summary["SOCAT_low"])
monthly_flux_summary["SOCAT_predictions_sigma_PgC_yr"] = 0.5*(monthly_flux_summary["SOCAT_predictions_high"]-monthly_flux_summary["SOCAT_predictions_low"])

monthly_flux_summary["floats_sigma_PgC_yr"] = 0.5*(monthly_flux_summary["floats_high"]-monthly_flux_summary["floats_low"])
monthly_flux_summary["floats_predictions_sigma_PgC_yr"] = 0.5*(monthly_flux_summary["floats_predictions_high"]-monthly_flux_summary["floats_predictions_low"])

# ============================================================
# 11) ANNUAL MEANS + MULTI-YEAR MEAN
# ============================================================

yearly_flux_summary = (
    monthly_flux_summary
    .groupby("year")[[
        "SOCAT","SOCAT_sigma_PgC_yr","SOCAT_predictions","SOCAT_predictions_sigma_PgC_yr",
        "floats","floats_sigma_PgC_yr","floats_predictions","floats_predictions_sigma_PgC_yr"
    ]]
    .mean()
    .reset_index()
)

mask = (yearly_flux_summary["year"]>=2003)&(yearly_flux_summary["year"]<=2024)
mean_vals = yearly_flux_summary.loc[mask].mean(numeric_only=True)

# ==================================================================
# ===================== APPENDED: COMBINED ONLY ======================
# ======================================================================

# ---------- LOAD COMBINED (UNWEIGHTED + WEIGHTED) ----------
predictions_combined_unweighted = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/pco2_prediction_matrix_SOCAT_AND_FLOATS_unweighted_MCensemble.csv"
)

predictions_combined_unweighted["month_center"] = pd.to_datetime(predictions_combined_unweighted["month_center"])
predictions_combined_unweighted = predictions_combined_unweighted[
    ~((predictions_combined_unweighted["month_center"].dt.year == 2023) &
      (predictions_combined_unweighted["month_center"].dt.month == 2))
]

predictions_combined_weighted = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/pco2_prediction_matrix_SOCAT_AND_FLOATS_weighted_MCensemble.csv"
)

predictions_combined_weighted["month_center"] = pd.to_datetime(predictions_combined_weighted["month_center"])
predictions_combined_weighted = predictions_combined_weighted[
    ~((predictions_combined_weighted["month_center"].dt.year == 2023) &
      (predictions_combined_weighted["month_center"].dt.month == 2))
]

combined_df = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/MODEL_SOCAT_AND_FLOATS_v2/processing_steps/pco2_all_splits_combined.csv",
    low_memory=False
)

# ---------- COMBINED OBS UNCERTAINTY ----------
combined_df_low  = combined_df.copy()
combined_df_high = combined_df.copy()
combined_df_low["pco2"]  -= combined_df_low["uncertainty_pco2"]
combined_df_high["pco2"] += combined_df_high["uncertainty_pco2"]

# ---------- COMBINED MODEL UNCERTAINTY (UNWEIGHTED + WEIGHTED) ----------
pred_combined_unweighted_low  = predictions_combined_unweighted.copy()
pred_combined_unweighted_high = predictions_combined_unweighted.copy()
sigma_model_combined_unweighted = predictions_combined_unweighted["ensemble_pco2_std"].values
pred_combined_unweighted_low["ensemble_pco2_pred"]  -= sigma_model_combined_unweighted
pred_combined_unweighted_high["ensemble_pco2_pred"] += sigma_model_combined_unweighted

pred_combined_weighted_low  = predictions_combined_weighted.copy()
pred_combined_weighted_high = predictions_combined_weighted.copy()
sigma_model_combined_weighted = predictions_combined_weighted["ensemble_pco2_std"].values
pred_combined_weighted_low["ensemble_pco2_pred"]  -= sigma_model_combined_weighted
pred_combined_weighted_high["ensemble_pco2_pred"] += sigma_model_combined_weighted

# ---------- COMBINED FLUX FUNCTIONS ----------
def flux_combined_obs(df):
    return flux_from_pco2(df, df["temperature"].values,
                           df["salinity"].values, df["pco2"].values)

def flux_combined_pred(df):
    return flux_from_pco2(df, df["temperature"].values,
                           df["salinity"].values, df["ensemble_pco2_pred"].values)

# ---------- PROCESS COMBINED OBS ----------
for df in [combined_df, combined_df_low, combined_df_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_combined_obs(df) * df["time"].dt.days_in_month

combined       = area_weighted_monthly_mean(bin_flux(combined_df,"latitude","longitude","time","flux_month"))
combined_low   = area_weighted_monthly_mean(bin_flux(combined_df_low,"latitude","longitude","time","flux_month"))
combined_high  = area_weighted_monthly_mean(bin_flux(combined_df_high,"latitude","longitude","time","flux_month"))

combined.columns      = ["year","month","date","combined"]
combined_low.columns  = ["year","month","date","combined_low"]
combined_high.columns = ["year","month","date","combined_high"]

# ---------- PROCESS COMBINED PRED (UNWEIGHTED) ----------
for df in [predictions_combined_unweighted, pred_combined_unweighted_low, pred_combined_unweighted_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_combined_pred(df) * df["time"].dt.days_in_month

pred_combined_unweighted       = area_weighted_monthly_mean(bin_flux(predictions_combined_unweighted,"latitude","longitude","time","flux_month"))
pred_combined_unweighted_low   = area_weighted_monthly_mean(bin_flux(pred_combined_unweighted_low,"latitude","longitude","time","flux_month"))
pred_combined_unweighted_high  = area_weighted_monthly_mean(bin_flux(pred_combined_unweighted_high,"latitude","longitude","time","flux_month"))

pred_combined_unweighted.columns      = ["year","month","date","combined_predictions_unweighted"]
pred_combined_unweighted_low.columns  = ["year","month","date","combined_predictions_unweighted_low"]
pred_combined_unweighted_high.columns = ["year","month","date","combined_predictions_unweighted_high"]

# ---------- PROCESS COMBINED PRED (WEIGHTED) ----------
for df in [predictions_combined_weighted, pred_combined_weighted_low, pred_combined_weighted_high]:
    df["time"] = pd.to_datetime(df["month_center"])
    df["flux_month"] = flux_combined_pred(df) * df["time"].dt.days_in_month

pred_combined_weighted       = area_weighted_monthly_mean(bin_flux(predictions_combined_weighted,"latitude","longitude","time","flux_month"))
pred_combined_weighted_low   = area_weighted_monthly_mean(bin_flux(pred_combined_weighted_low,"latitude","longitude","time","flux_month"))
pred_combined_weighted_high  = area_weighted_monthly_mean(bin_flux(pred_combined_weighted_high,"latitude","longitude","time","flux_month"))

pred_combined_weighted.columns      = ["year","month","date","combined_predictions_weighted"]
pred_combined_weighted_low.columns  = ["year","month","date","combined_predictions_weighted_low"]
pred_combined_weighted_high.columns = ["year","month","date","combined_predictions_weighted_high"]

# ---------- MERGE COMBINED INTO A NEW DF ----------
monthly_flux_summary_combined = (
    monthly_flux_summary
    .merge(combined,on=["year","month","date"])
    .merge(combined_low,on=["year","month","date"])
    .merge(combined_high,on=["year","month","date"])
    .merge(pred_combined_unweighted,on=["year","month","date"])
    .merge(pred_combined_unweighted_low,on=["year","month","date"])
    .merge(pred_combined_unweighted_high,on=["year","month","date"])
    .merge(pred_combined_weighted,on=["year","month","date"])
    .merge(pred_combined_weighted_low,on=["year","month","date"])
    .merge(pred_combined_weighted_high,on=["year","month","date"])
)

# ---------- SCALE COMBINED ONLY ----------
for col in [
    "combined","combined_low","combined_high",
    "combined_predictions_unweighted","combined_predictions_unweighted_low","combined_predictions_unweighted_high",
    "combined_predictions_weighted","combined_predictions_weighted_low","combined_predictions_weighted_high"
]:
    monthly_flux_summary_combined[col] = (
        monthly_flux_summary_combined[col] * 12 * OCEAN_AREA_M2 / 1e15
    )

# ---------- COMBINED SIGMA (UNWEIGHTED + WEIGHTED) ----------
monthly_flux_summary_combined["combined_sigma_PgC_yr"] = (
    0.5*(monthly_flux_summary_combined["combined_high"] -
         monthly_flux_summary_combined["combined_low"])
)

monthly_flux_summary_combined["combined_predictions_unweighted_sigma_PgC_yr"] = (
    0.5*(monthly_flux_summary_combined["combined_predictions_unweighted_high"] -
         monthly_flux_summary_combined["combined_predictions_unweighted_low"])
)

monthly_flux_summary_combined["combined_predictions_weighted_sigma_PgC_yr"] = (
    0.5*(monthly_flux_summary_combined["combined_predictions_weighted_high"] -
         monthly_flux_summary_combined["combined_predictions_weighted_low"])
)

# ---------- COMBINED ANNUALS (UNWEIGHTED + WEIGHTED) ----------
yearly_flux_summary_combined = (
    monthly_flux_summary_combined
    .groupby("year")[[
        "combined","combined_sigma_PgC_yr",
        "combined_predictions_unweighted","combined_predictions_unweighted_sigma_PgC_yr",
        "combined_predictions_weighted","combined_predictions_weighted_sigma_PgC_yr"
    ]]
    .mean()
    .reset_index()
)

mask_c = (yearly_flux_summary_combined["year"]>=2003)&(yearly_flux_summary_combined["year"]<=2024)
mean_vals_combined = yearly_flux_summary_combined.loc[mask_c].mean(numeric_only=True)
```

## Uncertainty difference from panel a to panel b

```python
import numpy as np
import pandas as pd

YEAR_MIN, YEAR_MAX = 2003, 2024

df = yearly_flux_summary.copy()

# --- restrict to analysis period ---
df = df[(df["year"] >= YEAR_MIN) & (df["year"] <= YEAR_MAX)].copy()

print(df.columns)  # quick sanity check
# ============================================================
# Helper: summarize annual ±1σ envelopes
# ============================================================

def summarize_sigma(df, label, sig_obs_col, sig_mod_col):
    tmp = df[["year", sig_obs_col, sig_mod_col]].copy()
    tmp = tmp.replace([np.inf, -np.inf], np.nan).dropna()
    tmp = tmp[(tmp[sig_obs_col] > 0) & (tmp[sig_mod_col] >= 0)]

    sigma_obs = tmp[sig_obs_col].values
    sigma_mod = tmp[sig_mod_col].values

    ratio = sigma_mod / sigma_obs
    pct_change = 100.0 * (sigma_mod - sigma_obs) / sigma_obs

    out = {
        "pair": label,
        "n_years": len(tmp),
        "median_sigma_obs": np.median(sigma_obs),
        "median_sigma_mod": np.median(sigma_mod),
        "median_pct_change": np.median(pct_change),
        "mean_sigma_obs": np.mean(sigma_obs),
        "mean_sigma_mod": np.mean(sigma_mod),
        "mean_pct_change": np.mean(pct_change),
    }
    return out

pairs = [
    ("SOCAT: obs → model",
     "SOCAT_sigma_PgC_yr", "SOCAT_predictions_sigma_PgC_yr"),

    ("FLOATS: obs → model",
     "floats_sigma_PgC_yr", "floats_predictions_sigma_PgC_yr"),

    ("COMBINED (weighted): obs → model",
     "combined_sigma_PgC_yr", "combined_predictions_weighted_sigma_PgC_yr"),
]

rows = []
for label, ocol, mcol in pairs:
    rows.append(summarize_sigma(df, label, ocol, mcol))

paper = pd.DataFrame(rows)

# ============================================================
# PRINT PAPER-READY NUMBERS
# ============================================================

print("\n=== Annual ±1σ envelope comparison (2003–2024) ===\n")
for _, r in paper.iterrows():
    direction = "reduction" if r["median_pct_change"] < 0 else "increase"
    print(
        f"{r['pair']} (n={int(r['n_years'])}): "
        f"median σobs={r['median_sigma_obs']:.2f}, "
        f"median σmodel={r['median_sigma_mod']:.2f} Pg C yr⁻¹; "
        f"median change={r['median_pct_change']:+.1f}% ({direction})"
    )

print("\n--- Full table ---")
print(paper.to_string(index=False))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# --- time axis ---
yearly_flux_summary["year_center"] = pd.to_datetime(
    yearly_flux_summary["year"].astype(str) + "-07-01"
)
t = yearly_flux_summary["year_center"]
xmin, xmax = t.min(), t.max()

# ---------- PANEL A (OBS) series (order matters) ----------
obs_series = [
    dict(key="SOCAT",   sigma="SOCAT_sigma_PgC_yr",   color="k",         label="SOCATv2025",          z=8, alpha=0.15),
    dict(key="floats",  sigma="floats_sigma_PgC_yr",  color="tab:blue",  label="BGC-Argo floats",     z=6, alpha=0.15),
    dict(key="combined",sigma="combined_sigma_PgC_yr",color="#A4D4B4",   label="SOCAT + FLOATS (obs)",z=4, alpha=0.25),
]

# ---------- PANEL B (MODELS) series (order matters) ----------
model_series = [
    dict(key="SOCAT_predictions", sigma="SOCAT_predictions_sigma_PgC_yr", color="k",
         label=r"Model$_{\mathrm{SOCATv2025}}$", z=8, alpha=0.15),
    dict(key="floats_predictions", sigma="floats_predictions_sigma_PgC_yr", color="tab:blue",
         label=r"Model$_{\mathrm{BGC\!-\!Argo\ floats}}$", z=6, alpha=0.15),
    dict(key="combined_predictions_unweighted", sigma="combined_predictions_unweighted_sigma_PgC_yr", color="#A4D4B4",
         label=r"Model$_{\mathrm{SOCAT+FLOATS}}$ (unweighted)", z=4, alpha=0.25),
    dict(key="combined_predictions_weighted", sigma="combined_predictions_weighted_sigma_PgC_yr", color="#2E8B57",
         label=r"Model$_{\mathrm{SOCAT+FLOATS}}$ (weighted)", z=2, alpha=0.18),
]

# ============================================================
# HELPER: plot one progressive step into a single-axis figure
# ============================================================
def plot_progressive(ax, df, time, series_subset, ylabel):
    handles, labels = [], []

    for s in series_subset:
        line, = ax.plot(
            time, df[s["key"]],
            linestyle="--", lw=1.5, color=s["color"],
            zorder=s["z"], label=s["label"]
        )
        ax.fill_between(
            time,
            df[s["key"]] - df[s["sigma"]],
            df[s["key"]] + df[s["sigma"]],
            color=s["color"], alpha=s["alpha"],
            zorder=s["z"] - 1
        )
        handles.append(line)
        labels.append(s["label"])

    ax.set_ylabel(ylabel)
    ax.axhline(0, color="k", lw=0.8)
    ax.grid(True)
    ax.set_xlim(xmin, xmax)
    ax.set_ylim(-3, 2)

    return handles, labels

# ============================================================
# LOOP 1: OBS figure only (Panel A)
# ============================================================
for step in range(1, len(obs_series) + 1):
    fig, ax = plt.subplots(figsize=(10, 5), dpi=300)

    handles, labels = plot_progressive(
        ax=ax,
        df=yearly_flux_summary,
        time=t,
        series_subset=obs_series[:step],
        ylabel="Observed CO$_2$ flux (Pg C yr$^{-1}$)",
    )
    ax.set_xlabel("Year")

    fig.legend(handles, labels, loc="lower center", ncol=3, frameon=False)
    fig.tight_layout(rect=[0, 0.12, 1, 1])

    fig.savefig(
        f"/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/figs/presentation/"
        f"OBS_progressive_step_{step}.png",
        dpi=300, bbox_inches="tight"
    )
    plt.close(fig)

# ============================================================
# LOOP 2: MODEL figure only (Panel B)
# ============================================================
for step in range(1, len(model_series) + 1):
    fig, ax = plt.subplots(figsize=(10, 5), dpi=300)

    handles, labels = plot_progressive(
        ax=ax,
        df=yearly_flux_summary,
        time=t,
        series_subset=model_series[:step],
        ylabel="Modeled CO$_2$ flux (Pg C yr$^{-1}$)",
    )
    ax.set_xlabel("Year")

    fig.legend(handles, labels, loc="lower center", ncol=2, frameon=False)
    fig.tight_layout(rect=[0, 0.12, 1, 1])

    fig.savefig(
        f"/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/figs/presentation/"
        f"MODEL_progressive_step_{step}.png",
        dpi=300, bbox_inches="tight"
    )
    plt.close(fig)
```

## GCB fCO₂-Based Southern Ocean fluxes (<30°S)

```python
import xarray as xr
import numpy as np
import pandas as pd
import glob

path_gcb = "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/GCB/Ocean/"
fco2_files = sorted(glob.glob(path_gcb + "GCB-2025_dataprod_*.nc"))

south_list = []
product_names = []
bad_files = []

for f in fco2_files:
    try:
        ds = xr.open_dataset(f)

        if "fgco2_reg" not in ds:
            raise ValueError("fgco2_reg missing")

        # SOUTH = region index 3 (<30°S)
        south = ds["fgco2_reg"][:, 3].to_series()

        # =============== FIX FOR JENA-MLS VS MONTHLY PRODUCTS ===============
        if len(south) == len(ds["time"]):
            # Normal monthly product → resample
            south.index = pd.to_datetime(ds["time"].values)
            south_ann = south.resample("YE").mean()
        else:
            # Already annual (e.g. JENA-MLS)
            start_year = int(ds["time"].dt.year.min())
            years = np.arange(start_year, start_year + len(south))
            south_ann = south.copy()
            south_ann.index = pd.to_datetime(years, format="%Y")

        # =============== APPLY GCB SOUTH RIVER CORRECTION ====================
        south_ann = south_ann + 0.09   # Pg C yr-1

        # =============== CONVERT TO YOUR SIGN CONVENTION ====================
        south_ann = -south_ann        # NEGATIVE = sink

        # =============== RESTRICT TO YOUR ANALYSIS PERIOD ===================
        south_ann = south_ann[
            (south_ann.index.year >= 2003) &
            (south_ann.index.year <= 2024)
        ]

        south_list.append(south_ann.values)
        product_names.append(ds.attrs.get("model_name", f.split("/")[-1]))

        print("Loaded:", f.split("/")[-1])

    except Exception as e:
        print("FAILED:", f.split("/")[-1])
        print("   Reason:", str(e))
        bad_files.append(f)

# ===================== BUILD FINAL DATASET =====================

years = south_ann.index.year

fco2_ds = xr.Dataset(
    data_vars=dict(
        fgco2_products=(["product", "year"], np.vstack(south_list))
    ),
    coords=dict(
        product=("product", product_names),
        year=("year", years)
    ),
    attrs=dict(
        description="GCB 2025 fCO2-based products, South <30°S, annual mean",
        time_period="2003–2024",
        units="Pg C yr-1",
        sign_convention="Negative = ocean CO2 sink",
        river_correction_applied="+0.09 Pg C yr-1 (South)"
    )
)

print("\nFINAL DATASET READY")
print("Products:", len(product_names))
print("Years:", years[0], "→", years[-1])
print("Failed files:", bad_files)
```

## Atmospheric inversions

```python
# ============================================================
# GCB2025 Atmospheric inversions: Southern Ocean (< -30°S)
# Annual regional flux in PgC/yr, with a time-varying ensemble:
# each inversion member is included only from its first "active"
# year onward (detected from non-zero regional annual flux).
#
# Output sign convention: whatever the file provides.
# (If you want "uptake positive", flip sign at the end.)
# NO river adjustment applied to atmospheric inversions.
# ============================================================

import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------
# 1) Load dataset
# ----------------------------
inv_file = (
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/GCB/Atmosphere/"
    "GCP2025_inversions_1x1_version1_0_20251031.nc"
)

ds = xr.open_dataset(inv_file, engine="h5netcdf")

# ----------------------------
# 2) Define region mask: lat < -30
# ----------------------------
mask_south = ds["latitude"] < -30

# ----------------------------
# 3) Convert monthly flux rate -> monthly PgC
#    f is [PgC m-2 yr-1] at monthly resolution
#    monthly PgC = f * area * (days_in_month / 365.25)
# ----------------------------
f = ds["ocean_flux_NOT_adjusted"]                  # (ensemble_member, time, lat, lon)
A = ds["cell_area"]                               # (lat, lon)
days = ds["time"].dt.days_in_month                # (time,)

f_month_PgC = f * A * (days / 365.25)

# ----------------------------
# 4) Integrate over region -> PgC/month
# ----------------------------
south_month_PgC = f_month_PgC.where(mask_south).sum(
    dim=("latitude", "longitude"),
    skipna=True
)

# ----------------------------
# 5) Aggregate to annual PgC/yr
# ----------------------------
south_annual_PgCyr = south_month_PgC.groupby("time.year").sum("time")
south_annual_PgCyr = south_annual_PgCyr.rename({"year": "year"})  # explicit

# ----------------------------
# 6) Decode inversion member names (optional but useful)
# ----------------------------
member_names = [
    "".join(ds["ensemble_member_name"].sel(ensemble_member=i).values)
    for i in ds["ensemble_member"].values
]
print("\nEnsemble members:")
for i, n in enumerate(member_names):
    print(f"{i:2d}  {n}")

# ----------------------------
# 7) Time-varying ensemble inclusion:
#    For each member, find first year when |annual flux| > eps.
#    Mask that member before that year, then compute mean/std.
# ----------------------------
eps = 0.05  # PgC/yr threshold to detect "active" (tune if needed)

member_ann = south_annual_PgCyr  # (ensemble_member, year)

active = np.abs(member_ann) > eps  # True where member-year is "active"

# idxmax gives first coordinate of max (True > False); but if all False,
# idxmax returns the first year, so we must guard against "never active".
any_active = active.any(dim="year")

start_year = active.where(active).idxmax(dim="year")
start_year = start_year.where(any_active)  # members never active -> NaN

# Mask values prior to each member's start year
member_masked = member_ann.where(member_ann["year"] >= start_year)

# Ensemble stats ignoring NaNs
ens_mean = member_masked.mean("ensemble_member", skipna=True)
ens_std  = member_masked.std("ensemble_member", skipna=True)

# Number of contributing members each year
n_members = member_masked.notnull().sum("ensemble_member")

# ----------------------------
# 8) Restrict to your comparison period
# ----------------------------
year_min, year_max = 2003, 2024
ens_mean = ens_mean.sel(year=slice(year_min, year_max))
ens_std  = ens_std.sel(year=slice(year_min, year_max))
n_members = n_members.sel(year=slice(year_min, year_max))

member_masked_period = member_masked.sel(year=slice(year_min, year_max))

# If you want uptake positive (GCB Fig.14 convention), uncomment:
# ens_mean = -ens_mean
# (std unchanged)

# ----------------------------
# 9) Diagnostics: print start years and contributing count
# ----------------------------
print("\nDetected start year per member (eps = %.3f PgC/yr):" % eps)
for i, (n, y) in enumerate(zip(member_names, start_year.values)):
    y_str = "None" if np.isnan(y) else str(int(y))
    print(f"{i:2d}  {n:25s}  start_year={y_str}")

print("\nMembers contributing by year (after masking):")
for y, nm in zip(n_members["year"].values, n_members.values):
    print(f"{int(y)}: {int(nm)}")

# ----------------------------
# 10) Plot: (a) per-member (masked) (b) ensemble mean +/- 1σ
# ----------------------------
time = pd.to_datetime(ens_mean["year"].values.astype(str) + "-07-01")

plt.figure(figsize=(11, 4), dpi=160)
for i in member_masked_period["ensemble_member"].values:
    plt.plot(
        member_masked_period["year"].values,
        member_masked_period.sel(ensemble_member=i).values,
        alpha=0.5,
        lw=1.2
    )
plt.axhline(0, color="k", lw=0.8)
plt.title("Southern Ocean (< -30°) annual flux per inversion member (masked before start year)")
plt.xlabel("Year")
plt.ylabel("Regional CO$_2$ flux (Pg C yr$^{-1}$)")
plt.grid(True, alpha=0.3)
plt.show()

plt.figure(figsize=(11, 4), dpi=160)
plt.plot(time, ens_mean.values, lw=2.5, label="Inversion ensemble mean")
plt.fill_between(
    time,
    (ens_mean - ens_std).values,
    (ens_mean + ens_std).values,
    alpha=0.2,
    label=r"$\pm 1\sigma$"
)
plt.axhline(0, color="k", lw=0.8)
plt.title("Southern Ocean (< -30°) atmospheric inversions: mean ± 1σ (time-varying ensemble)")
plt.xlabel("Year")
plt.ylabel("Regional CO$_2$ flux (Pg C yr$^{-1}$)")
plt.grid(True, alpha=0.3)

# Optional: show number of contributing members on a secondary axis
ax = plt.gca()
ax2 = ax.twinx()
ax2.plot(time, n_members.values, lw=1.5, alpha=0.6)
ax2.set_ylabel("N ensemble members")

plt.legend(loc="best")
plt.show()

# ----------------------------
# 11) Final outputs for your main figure
# ----------------------------
# ens_mean: (year,) annual PgC/yr
# ens_std : (year,) annual PgC/yr
# n_members: (year,) number of active members
#
# Convert to arrays if needed:
inv_years = ens_mean["year"].values
inv_mean = ens_mean.values
inv_std  = ens_std.values

print("\nReady for plotting:")
print("inv_years shape:", inv_years.shape)
print("inv_mean shape :", inv_mean.shape)
print("inv_std shape  :", inv_std.shape)
```

```python
# ============================================================
# SUMMARY BAR PLOT (OBS + MODELS + GCB)
# mean bars + 95% CI errorbars at bottom
# text = mean ± σ_mean
#
# + PRINT TABLE S2 (mean, sigma_mean, CI95, P(mean<0))
# ============================================================

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

YEAR_MIN, YEAR_MAX = 2003, 2024
N_MC = 200000
SEED = 42

# ============================================================
# Monte Carlo helper
# ============================================================

def mc_stats(Fy, sy, N=200000, seed=42):
    """
    Returns:
      mean_est   : multi-year mean of annual means
      sigma_mean : std of MC distribution of the multi-year mean
      ci95       : (2.5%, 97.5%) of MC distribution of the multi-year mean
      p_neg      : P(mean < 0) from MC draws (useful for Table S2)
    """
    Fy = np.asarray(Fy, dtype=float)
    sy = np.asarray(sy, dtype=float)

    m = np.isfinite(Fy) & np.isfinite(sy)
    Fy, sy = Fy[m], sy[m]
    if Fy.size == 0:
        return np.nan, np.nan, (np.nan, np.nan), np.nan

    rng = np.random.default_rng(seed)
    draws = rng.normal(loc=Fy, scale=sy, size=(N, Fy.size))
    means = draws.mean(axis=1)

    mean_est = Fy.mean()
    sigma_mean = means.std(ddof=0)
    ci95 = np.quantile(means, [0.025, 0.975])
    p_neg = (means < 0).mean()

    return mean_est, sigma_mean, (ci95[0], ci95[1]), p_neg


# ============================================================
# Subset to 2003–2024
# ============================================================

df = yearly_flux_summary.copy()
df = df[(df["year"] >= YEAR_MIN) & (df["year"] <= YEAR_MAX)]

dfc = yearly_flux_summary_combined.copy()
dfc = dfc[(dfc["year"] >= YEAR_MIN) & (dfc["year"] <= YEAR_MAX)]

fco2_plot = fco2_ds.sel(year=slice(YEAR_MIN, YEAR_MAX))
gobm_plot = gobm_ds.sel(year=slice(YEAR_MIN, YEAR_MAX))

mask_inv = (ens_mean["year"] >= YEAR_MIN) & (ens_mean["year"] <= YEAR_MAX)
inv_mean = ens_mean.values[mask_inv]
inv_std  = ens_std.values[mask_inv]


# ============================================================
# BUILD ROWS IN NEW ORDER
# Each row: (label, mean, sigma_mean, (ci_low, ci_high), p_neg, color)
# ============================================================

rows = []

# ---- SOCAT ----
m, s, ci, pneg = mc_stats(df["SOCAT"].values, df["SOCAT_sigma_PgC_yr"].values, N=N_MC, seed=SEED)
rows.append((r"OBS$_{\mathrm{SOCAT}}$", m, s, ci, pneg, "#929591"))

m, s, ci, pneg = mc_stats(df["SOCAT_predictions"].values,
                          df["SOCAT_predictions_sigma_PgC_yr"].values,
                          N=N_MC, seed=SEED)
rows.append((r"Model$_{\mathrm{SOCAT}}$", m, s, ci, pneg, "#929591"))

# ---- FLOATS ----
m, s, ci, pneg = mc_stats(df["floats"].values, df["floats_sigma_PgC_yr"].values, N=N_MC, seed=SEED)
rows.append((r"OBS$_{\mathrm{BGC{-}Argo\ floats}}$", m, s, ci, pneg, "tab:blue"))

m, s, ci, pneg = mc_stats(df["floats_predictions"].values,
                          df["floats_predictions_sigma_PgC_yr"].values,
                          N=N_MC, seed=SEED)
rows.append((r"Model$_{\mathrm{BGC{-}Argo\ floats}}$", m, s, ci, pneg, "tab:blue"))

# ---- COMBINED ----
m, s, ci, pneg = mc_stats(dfc["combined"].values, dfc["combined_sigma_PgC_yr"].values, N=N_MC, seed=SEED)
rows.append((r"OBS$_{\mathrm{SOCAT+BGC{-}Argo\ floats}}$", m, s, ci, pneg, "#2E8B57"))

m, s, ci, pneg = mc_stats(dfc["combined_predictions_weighted"].values,
                          dfc["combined_predictions_weighted_sigma_PgC_yr"].values,
                          N=N_MC, seed=SEED)
rows.append((r"Model$_{\mathrm{SOCAT+BGC{-}Argo\ floats\ (weighted)}}$", m, s, ci, pneg, "#2E8B57"))

# ---- GCB ----
m, s, ci, pneg = mc_stats(inv_mean, inv_std, N=N_MC, seed=SEED)
rows.append(("GCB25 atmospheric inversions", m, s, ci, pneg, "#6A0DAD"))

Fy = fco2_plot["fgco2_products"].mean("product").values
sy = fco2_plot["fgco2_products"].std("product").values
m, s, ci, pneg = mc_stats(Fy, sy, N=N_MC, seed=SEED)
rows.append(("GCB25 $f\\mathrm{CO}_2$ products", m, s, ci, pneg, "#2EC4B6"))

m, s, ci, pneg = mc_stats(gobm_plot["gobm_mean"].values,
                          gobm_plot["gobm_std"].values,
                          N=N_MC, seed=SEED)
rows.append(("GCB25 GOBMs (Sim A)", m, s, ci, pneg, "k"))


# ============================================================
# PRINT TABLE S2 CONTENT (publication-ready)
# ============================================================

print("\n=== Table S2. Significance of multi-year mean Southern Ocean CO2 flux (2003–2024, south of 30°S) ===")
print("Reported as mean ± σ_mean, with 95% CI from Monte Carlo propagation of annual uncertainties.\n")

table_rows = []
for label, mean_est, sigma_mean, (ci_l, ci_u), pneg, color in rows:
    sig = "YES" if (ci_l < 0) and (ci_u < 0) else "NO"
    table_rows.append({
        "Dataset": label,
        "Mean (Pg C yr-1)": mean_est,
        "σ_mean (Pg C yr-1)": sigma_mean,
        "95% CI low (Pg C yr-1)": ci_l,
        "95% CI high (Pg C yr-1)": ci_u,
        "P(mean < 0)": pneg,
        "Significant sink?": sig
    })
    # nice human-readable printout
    print(f"{label}")
    print(f"  mean ± σ_mean = {mean_est: .2f} ± {sigma_mean:.2f} Pg C yr⁻¹")
    print(f"  95% CI        = {ci_l: .2f} to {ci_u:.2f} Pg C yr⁻¹")
    print(f"  P(mean < 0)   = {pneg:.4f}")
    print(f"  Significant sink? {sig}\n")

df_S2 = pd.DataFrame(table_rows)


# ============================================================
# ARRAYS FOR PLOTTING (mirror order)
# ============================================================

labels_plot = [r[0] for r in rows][::-1]
means_plot  = np.array([r[1] for r in rows], dtype=float)[::-1]
sigmas_plot = np.array([r[2] for r in rows], dtype=float)[::-1]
ci_low      = np.array([r[3][0] for r in rows], dtype=float)[::-1]
ci_high     = np.array([r[3][1] for r in rows], dtype=float)[::-1]
colors      = [r[5] for r in rows][::-1]

err_left  = means_plot - ci_low
err_right = ci_high - means_plot


# ============================================================
# PLOT
# ============================================================

fig, ax = plt.subplots(figsize=(9, 6.0), dpi=300)
ypos = np.arange(len(labels_plot))
bar_height = 0.6

ax.barh(ypos, means_plot, color=colors, alpha=0.75, height=bar_height)

# 95% CI bottom bars
for i, (mval, el, er) in enumerate(zip(means_plot, err_left, err_right)):
    ax.errorbar(
        mval,
        i - bar_height/2,
        xerr=np.array([[el], [er]]),
        fmt="none",
        ecolor="black",
        elinewidth=1.8,
        capsize=4,
        zorder=5,
    )

# Labels
for i, (label, mval, smean) in enumerate(zip(labels_plot, means_plot, sigmas_plot)):
    ax.text(-0.02, i, label, va="center", ha="right", color="white", fontsize=11)
    ax.text(mval - 0.02, i, f"{mval:.2f} ± {smean:.2f}",
            va="center", ha="right", fontsize=9.5, color="black")

# Formatting
ax.set_yticks([])
ax.set_xlabel("Mean CO$_2$ uptake (Pg C yr$^{-1}$)", fontsize=11)
ax.set_xlim(-2.0, 0)
ax.grid(axis="x", alpha=0.3)

plt.tight_layout()

fig.savefig(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/figs/"
    "MEAN_UPTAKE_OBS_MODELS_GCB_ORDERED.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# PRINT NUMERICAL SUMMARY (same values used in bars)
# ============================================================

print("\n=== Multi-year mean Southern Ocean CO₂ uptake (2003–2024) ===")
print("Reported as: mean ± σ_mean   (95% CI of multi-year mean)\n")

for label, mean_est, sigma_mean, (ci_l, ci_u), pneg, color in rows:
    print(f"{label}")
    print(f"  mean = {mean_est:6.2f} Pg C yr⁻¹")
    print(f"  σ_mean = {sigma_mean:5.2f}")
    print(f"  95% CI = [{ci_l:5.2f}, {ci_u:5.2f}]")
    print(f"  P(mean < 0) = {pneg:.4f}\n")
```

```python
from scipy.stats import linregress
import numpy as np

# ============================================================
# What this script reports (YOUR requested convention)
# ------------------------------------------------------------
# - We fit a linear trend of FLUX vs YEAR within each window.
# - Flux convention: negative = uptake.
# - We REPORT the flux trend dF/dt in units of Pg C yr^-1 decade^-1.
#   (This is just the fitted per-year slope expressed per decade.)
# - We DO NOT multiply by -1 anywhere (no "sink-growth" sign flip).
# - z-scores are ONLY computed for 2003–2018 and compare to Hauck et al.
#   Hauck reports sink-growth (+ = stronger sink), so we convert Hauck to
#   flux-trend convention by flipping the sign of the reference.
# ============================================================

# ==========================
# Helpers
# ==========================
def fit_trend(years, flux):
    """Return slope, stderr, p for flux = a*year + b. slope is per YEAR."""
    slope, _, _, p, stderr = linregress(years.astype(float), flux.astype(float))
    return slope, stderr, p

def summarize_window_flux_trend(label, years, flux, y0, y1, ref_val=None, ref_unc=None):
    """
    Fit trend over [y0, y1] and report:
      - trend_dec: dF/dt in Pg C yr^-1 decade^-1  (NEGATIVE => strengthening uptake)
      - trend_dec_se: SE of trend_dec
      - p: p-value of slope (same regardless of unit scaling)
      - deltaF: total change in flux across window (Pg C yr^-1), using fitted slope
      - z: optional z-score vs reference (only when ref provided)
    """
    m = (years >= y0) & (years <= y1)
    yrs = years[m]
    y   = flux[m]

    slope, stderr, p = fit_trend(yrs, y)  # slope per year: Pg C yr^-2

    # Express slope per decade (unit conversion; no sign flip)
    trend_dec    = slope * 10.0
    trend_dec_se = stderr * 10.0

    # Total change in flux across the window (Pg C yr^-1)
    deltaF = slope * (y1 - y0)

    out = {
        "label": label,
        "y0": y0, "y1": y1,
        "trend_dec": trend_dec,
        "trend_dec_se": trend_dec_se,
        "p": p,
        "deltaF": deltaF,
        "n": int(np.sum(m)),
        "ref_val": ref_val,
        "ref_unc": ref_unc,
        "z": None
    }

    if (ref_val is not None) and (ref_unc is not None):
        out["z"] = (trend_dec - ref_val) / np.sqrt(trend_dec_se**2 + ref_unc**2)

    return out

def fmt_sci(x):
    """Format p-values like 5.74 × 10⁻⁶ (ASCII-friendly)."""
    if x is None:
        return "NA"
    if x == 0:
        return "0"
    exp = int(np.floor(np.log10(abs(x))))
    mant = x / (10**exp)
    return f"{mant:.2f} × 10{exp:+d}".replace("+", "⁺").replace("-", "⁻")

def cell(trend, se, p, z):
    """Format as: +0.493 ± 0.070 (5.74 × 10⁻⁶; +5.04) ; z=NA if missing."""
    ztxt = f"{z:+.2f}" if z is not None else "NA"
    return f"{trend:+.3f} ± {se:.3f} ({fmt_sci(p)}; {ztxt})"

# ==========================
# Windows
# ==========================
windows = {
    "2003–2018 (Hauck et al., 2023a)": (2003, 2018),
    "2003–2012": (2003, 2012),
    "2015–2024": (2015, 2024),
}

# ==========================
# Hauck et al. (2023a) references
# Hauck reports SINK-GROWTH (positive = stronger sink) in Pg C yr^-1 decade^-1.
# We are reporting FLUX-TREND dF/dt (negative = stronger sink),
# so reference values must be SIGN-FLIPPED.
# ==========================
hauck_pco2_sinkgrowth, hauck_pco2_unc = 0.26, 0.06  # reported as sink-growth
hauck_gobm_sinkgrowth, hauck_gobm_unc = 0.11, 0.03  # reported as sink-growth

hauck_pco2_fluxtrend = -hauck_pco2_sinkgrowth  # convert to flux-trend convention
hauck_gobm_fluxtrend = -hauck_gobm_sinkgrowth  # convert to flux-trend convention

# ==========================
# Series getters (as in your script)
# Assumes these exist: yearly_flux_summary, fco2_ds, gobm_ds
# ==========================
def df_series(df, y0, y1, col):
    m = (df["year"] >= y0) & (df["year"] <= y1)
    return df.loc[m, "year"].values, df.loc[m, col].values

def gcb_fco2_series(y0, y1):
    sub = fco2_ds.sel(year=slice(y0, y1))
    yrs = sub["year"].values
    y   = sub["fgco2_products"].mean("product").values
    return yrs, y

def gcb_gobm_series(y0, y1):
    sub = gobm_ds.sel(year=slice(y0, y1))
    yrs = sub["year"].values
    y   = sub["gobm_mean"].values
    return yrs, y

# ==========================
# Your model columns
# ==========================
my_models = {
    "Model$_{\\mathrm{SOCATv2025}}$": "SOCAT_predictions",
    "Model$_{\\mathrm{BGC\\text{-}Argo\\ floats}}$": "floats_predictions",
    "Model$_{\\mathrm{SOCATv2025+BGC\\text{-}Argo\\ floats}}$ (weighted)": "combined_predictions_weighted",
}

# ==========================
# Build the EXACT table you want (3 rows × 3 windows)
# ==========================
table = {name: {} for name in my_models.keys()}

for win_name, (y0, y1) in windows.items():
    for label, col in my_models.items():
        yrs, y = df_series(yearly_flux_summary, y0, y1, col)

        # Only compute z-scores for 2003–2018 vs Hauck pCO2-products reference
        if y0 == 2003 and y1 == 2018:
            ref_val, ref_unc = hauck_pco2_fluxtrend, hauck_pco2_unc
        else:
            ref_val, ref_unc = None, None

        r = summarize_window_flux_trend(label, yrs, y, y0, y1, ref_val=ref_val, ref_unc=ref_unc)
        table[label][win_name] = cell(r["trend_dec"], r["trend_dec_se"], r["p"], r["z"])

# ==========================
# Print as a Word-friendly table (TSV-like)
# ==========================
cols = list(windows.keys())
print("Dataset\t" + "\t".join(cols))
for label in my_models.keys():
    print(label + "\t" + "\t".join(table[label][c] for c in cols))

# --------------------------
# Notes:
# - Negative values here mean flux becomes more negative -> uptake strengthens.
# - If you want to compare to Hauck in sink-growth units instead, don't sign-flip the reference.
# --------------------------
```

```python
# ============================================================
# SEASONAL MAPS (4 PANELS) — EXACT SAME FIGURE STYLE AS YOUR TEMPLATE
# ONLY DIFFERENCE: THE DATA PLOTTED = (SOCAT predictions − COMBINED weighted predictions)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.colors import Normalize

# ============================================================
# 1) LOAD PREDICTIONS (SOCAT + COMBINED WEIGHTED)
# ============================================================

predictions_socat = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/"
    "pco2_prediction_matrix_SOCAT_only_MCensemble.csv"
)

predictions_combined_weighted = pd.read_csv(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/matrix_predictions/"
    "pco2_prediction_matrix_SOCAT_AND_FLOATS_weighted_MCensemble.csv"
)

for df in (predictions_socat, predictions_combined_weighted):
    df["month_center"] = pd.to_datetime(df["month_center"])
    df.drop(
        df[(df["month_center"].dt.year == 2023) &
           (df["month_center"].dt.month == 2)].index,
        inplace=True
    )

# ============================================================
# 2) ADD DJF/MAM/JJA/SON SEASON LABELS (TO MATCH YOUR TEMPLATE)
# ============================================================

def assign_season(month):
    if month in [12, 1, 2]:
        return "DJF"
    elif month in [3, 4, 5]:
        return "MAM"
    elif month in [6, 7, 8]:
        return "JJA"
    else:
        return "SON"

for df in (predictions_socat, predictions_combined_weighted):
    df["month"] = df["month_center"].dt.month
    df["season"] = df["month"].apply(assign_season)

# ============================================================
# 3) SEASONAL MEAN PER GRID CELL (AUTO-DETECT LAT/LON)
# ============================================================

def seasonal_mean_pred(df, pred_col="ensemble_pco2_pred"):
    """
    Seasonal mean prediction per season per grid cell.
    Auto-detects:
      - latitude/longitude
      - OR lat_center/lon_center
    Returns columns: season, latitude, longitude, pco2_pred
    """
    if {"latitude", "longitude"}.issubset(df.columns):
        lat_col, lon_col = "latitude", "longitude"
    elif {"lat_center", "lon_center"}.issubset(df.columns):
        lat_col, lon_col = "lat_center", "lon_center"
    else:
        raise KeyError("No valid latitude/longitude columns found in dataframe.")

    out = (
        df.groupby(["season", lat_col, lon_col])[pred_col]
          .mean()
          .reset_index()
          .rename(columns={lat_col: "latitude", lon_col: "longitude", pred_col: "pco2_pred"})
    )
    return out

seasonal_socat = seasonal_mean_pred(predictions_socat)
seasonal_cw    = seasonal_mean_pred(predictions_combined_weighted)

# ============================================================
# 4) BUILD THE DATA TO PLOT: SOCAT − COMBINED WEIGHTED
# (this is the ONLY difference vs your template: variable/colormapping)
# ============================================================

df_seasonal = seasonal_socat.merge(
    seasonal_cw,
    on=["season", "latitude", "longitude"],
    suffixes=("_socat", "_cw"),
    how="inner"
)

df_seasonal["delta_pco2"] = df_seasonal["pco2_pred_socat"] - df_seasonal["pco2_pred_cw"]


```

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.colors import Normalize, ListedColormap

# ============================================================
# LOAD DATA
# ============================================================

fronts = xr.load_dataset(
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/fronts/62985.nc"
)

# Monthly sea-ice climatology DataArray (dims: month, yc, xc; units: percent)
seaice = xr.load_dataset("/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/fronts/sic_climatology_stats_sh_1991-2020.nc")
seaice_mmean = seaice.mmean


# ============================================================
# HELPERS: Orsi fronts plotting (handles 0..360 lon, NaNs, dateline jumps)
# ============================================================

def lon360_to_lon180(lon):
    lon = np.asarray(lon)
    return ((lon + 180) % 360) - 180

def plot_front_line(ax, lon, lat, *, jump_deg=20, **plot_kwargs):
    """
    Plot a front polyline safely on a global/polar map:
      - drops NaNs
      - converts lon 0..360 -> -180..180
      - splits line where longitude jumps (dateline / gaps) to avoid long chords
    """
    lon = np.asarray(lon).astype(float)
    lat = np.asarray(lat).astype(float)

    m = np.isfinite(lon) & np.isfinite(lat)
    lon = lon[m]
    lat = lat[m]

    lon = lon360_to_lon180(lon)
    if lon.size < 2:
        return

    dlon = np.abs(np.diff(lon))
    breaks = np.where(dlon > jump_deg)[0] + 1
    idxs = np.concatenate(([0], breaks, [lon.size]))

    for a, b in zip(idxs[:-1], idxs[1:]):
        if b - a < 2:
            continue
        ax.plot(
            lon[a:b], lat[a:b],
            transform=ccrs.PlateCarree(),
            **plot_kwargs
        )


# ============================================================
# HELPERS: Sea-ice seasonal MAX mask (ice-only fill, no contours)
# ============================================================

SEASON_TO_MONTHS = {
    "DJF": [12, 1, 2],
    "MAM": [3, 4, 5],
    "JJA": [6, 7, 8],
    "SON": [9, 10, 11],
}

def seasonal_max_from_monthly_clim(seaice_mmean, season):
    """
    Returns 2D seasonal MAX (grid-cell-wise) across the 3 months in the season.
    seaice_mmean values are percent (0..100).
    """
    months = SEASON_TO_MONTHS[season]
    return seaice_mmean.sel(month=months).max("month", skipna=True)

ICE_CMAP = ListedColormap(["#601ef9"])  # soft purple

def plot_ice_mask_seasonal_max_iceonly(
    ax,
    seaice_mmean,
    season,
    *,
    threshold_percent=15.0,
    alpha=0.35,
    zorder=12
):
    """
    Plot ONLY the ice cells (>= threshold_percent) as a yellow semi-transparent mask.
    Non-ice is NaN => fully transparent (avoids the 'purple polygon' footprint).
    Uses the seasonal MAX across the 3 months.
    """
    sic_max = seasonal_max_from_monthly_clim(seaice_mmean, season)

    lon2d = sic_max["lon"].values
    lat2d = sic_max["lat"].values

    ice = (sic_max.values >= threshold_percent)
    ice_only = np.where(ice, 1.0, np.nan)  # only ice plotted

    ax.pcolormesh(
        lon2d, lat2d, ice_only,
        transform=ccrs.PlateCarree(),
        shading="auto",
        cmap=ICE_CMAP,
        vmin=0, vmax=1,
        alpha=alpha,
        zorder=zorder
    )


# ============================================================
# PLOTTING FUNCTION (4 SEASONS)
# ============================================================

def plot_seasonal_difference_maps(
    df_seasonal,
    title_prefix,
    output_path,
    fronts=None,
    seaice_mmean=None,
    ice_threshold_percent=15.0,
    ice_alpha=0.35
):

    # Southern Hemisphere season order + labels
    season_order  = ["DJF", "MAM", "JJA", "SON"]
    season_labels = {
        "DJF": "Summer",
        "MAM": "Autumn",
        "JJA": "Winter",
        "SON": "Spring",
    }

    panel_letters = ["a", "b", "c", "d"]

    vmax = np.nanpercentile(np.abs(df_seasonal["delta_pco2"].values), 99)
    vmin = -vmax
    norm = Normalize(vmin=vmin, vmax=vmax)
    cmap = "RdBu_r"
    true_max = np.nanmax(df_seasonal["delta_pco2"].values)
    true_min = np.nanmin(df_seasonal["delta_pco2"].values)

    fig, axs = plt.subplots(
        2, 2,
        figsize=(10, 10),
        dpi=300,
        subplot_kw={"projection": ccrs.SouthPolarStereo()}
    )

    fig.subplots_adjust(hspace=0.08, wspace=0.08)
    axs = axs.flatten()

    for i, (ax, season) in enumerate(zip(axs, season_order)):

        ax.set_extent([-180, 180, -90, -30], crs=ccrs.PlateCarree())

        # Base map
        ax.coastlines(resolution="110m", linewidth=0.6)
        ax.add_feature(cfeature.LAND, facecolor="lightgray", zorder=0)
        ax.gridlines(draw_labels=False, linewidth=0.2, color="gray", alpha=0.5)

        df_s = df_seasonal[df_seasonal["season"] == season]

        # Main scatter
        sc = ax.scatter(
            df_s["longitude"],
            df_s["latitude"],
            c=df_s["delta_pco2"],
            s=10,
            cmap=cmap,
            norm=norm,
            transform=ccrs.PlateCarree(),
            zorder=2
        )

        # ------------------------------------------------------------
        # OVERLAY: Sea-ice presence mask (JJA + MAM + SON)
        # ------------------------------------------------------------
        if seaice_mmean is not None and season in {"MAM", "JJA", "SON"}:
            plot_ice_mask_seasonal_max_iceonly(
                ax,
                seaice_mmean,
                season,
                threshold_percent=ice_threshold_percent,
                alpha=ice_alpha,
                zorder=12
            )
        # ------------------------------------------------------------
        # OVERLAY: Orsi Polar Front (PF)
        # ------------------------------------------------------------
        if fronts is not None:
            plot_front_line(
                ax,
                fronts["LonPF"].values,
                fronts["LatPF"].values,
                jump_deg=20,
                linewidth=1.0,
                linestyle="--",
                color="k",
                zorder=15
            )

        # Features on top
        ax.add_feature(cfeature.LAND, facecolor="none", edgecolor="none", zorder=8)

        ax.gridlines(
            draw_labels=False,
            linewidth=0.2,
            color="gray",
            alpha=0.5,
            zorder=9
        )

        ax.coastlines(resolution="110m", linewidth=0.6, zorder=10)

        # Panel label
        ax.text(
            0, 1.05,
            f"{panel_letters[i]}) {season_labels[season]}",
            transform=ax.transAxes,
            ha="left",
            va="top",
            fontsize=11,
        )

    # ============================================================
    # COLORBAR
    # ============================================================
    cax = fig.add_axes([0.215, 0.06, 0.60, 0.018])

    cbar = fig.colorbar(sc, cax=cax, orientation="horizontal", extend="both")
    cbar.set_label(
        r"$\Delta p$CO$_2$ (µatm)  [Model$_{\mathrm{SOCAT}}$ − Model$_{\mathrm{SOCAT+FLOATS}}$ (weighted)]"
    )

    cbar.ax.text(
        -0.07, 0.45, f"{true_min:.1f}",
        transform=cbar.ax.transAxes, va="center", ha="right", fontsize=9
    )
    cbar.ax.text(
        1.07, 0.45, f"{true_max:.1f}",
        transform=cbar.ax.transAxes, va="center", ha="left", fontsize=9
    )

    plt.savefig(output_path, dpi=300, bbox_inches="tight")
    plt.show()


# ============================================================
# CALL
# ============================================================

plot_seasonal_difference_maps(
    df_seasonal,
    "SOCAT − Combined weighted predictions",
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/figs/"
    "SOCAT_minus_COMBINEDweighted_seasonal_difference_maps.png",
    fronts=fronts,
    seaice_mmean=seaice_mmean,
    ice_threshold_percent=15.0,  # 15% is the standard "extent" threshold
    ice_alpha=0.35               
)
```
