```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import pandas as pd

# --- Only needed for the ocean mask ---
from cartopy.io import shapereader
import shapely.geometry as sgeom
import shapely.prepared as prep
try:
    # Shapely ≥2.0 (and most 1.8+) puts unary_union in shapely.ops
    from shapely.ops import unary_union
except ImportError:
    # Fallback for very old Shapely
    from shapely.geometry import unary_union


# =========================
# User-configurable inputs
# =========================

# Prediction matrix (physics/ML fields on 1° monthly grid)
MATRIX_CSV = ("/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/PREDICTION_MATRIX_v2/processing_steps/ARMOR3D_TSMLD_SouthernOcean_1997_2024_monthly_1deg_feature_engineered_monthly_1deg.csv")

# SOCAT monthly 1° feature-engineered grid
SOCAT_CSV = ("/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/MODEL_SOCAT_ONLY_OBS_v2/processing_steps/SOCATv2025_SO_feature_engineered_monthly_1deg_pCO2.csv")

# BGC-Argo monthly 1° feature-engineered grid
FLOATS_CSV = ("/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/MODEL_FLOATS_v2/processing_steps/BGC-Argo_SO_feature_engineered_monthly_1deg.csv")

# Time range (years only now, since data are regular monthly)
START_YEAR = 2003
END_YEAR   = 2024
NUM_YEARS  = END_YEAR - START_YEAR + 1
NUM_MONTHS = NUM_YEARS * 12

# Spatial domain (1° grid cells, integer bins)
LAT_RANGE  = range(-90, -29)   # -90..-30 inclusive
LON_RANGE  = range(-180, 180)  # -180..179 inclusive


# =========================
# Helpers
# =========================
def build_ocean_cell_mask(lat_range, lon_range):
    """
    Return a set of (lat, lon) integer cell indices that are ocean,
    using 1° cell-center point test against Natural Earth land polygons.
    """
    land_shp = shapereader.natural_earth(
        resolution='50m',
        category='physical',
        name='land'
    )
    land_geoms = list(shapereader.Reader(land_shp).geometries())
    land_prep = prep.prep(unary_union(land_geoms))

    def is_ocean_cell(lat_cell, lon_cell):
        # Test the 1° cell center (lon+0.5, lat+0.5)
        return not land_prep.contains(
            sgeom.Point(lon_cell + 0.5, lat_cell + 0.5)
        )

    return {
        (lat, lon)
        for lat in lat_range
        for lon in lon_range
        if is_ocean_cell(lat, lon)
    }


def get_season(month: int) -> str:
    if month in (12, 1, 2): return "Summer"
    if month in (3, 4, 5):  return "Autumn"
    if month in (6, 7, 8):  return "Winter"
    return "Spring"


def prep_bins_ocean_only(
    df: pd.DataFrame,
    ocean_cells: set,
    lat_col: str = 'latitude',
    lon_col: str = 'longitude'
) -> pd.DataFrame:
    """
    For *already gridded monthly* data:
    - ensure year, month, lat, lon exist
    - cast lat/lon to int cell indices
    - keep only ocean cells
    """
    # Require year, month, lat, lon
    df = df.dropna(subset=[lat_col, lon_col, 'year', 'month']).copy()

    # Integer 1° cells
    df['latitude']  = df[lat_col].astype(int)
    df['longitude'] = df[lon_col].astype(int)

    # Keep only ocean cells
    ll = list(zip(df['latitude'], df['longitude']))
    df = df[[cell in ocean_cells for cell in ll]]

    return df


def compute_total_coverage_count(df: pd.DataFrame) -> int:
    """Count unique (year, month, lat, lon) bins."""
    unique_bins = df[['year', 'month', 'latitude', 'longitude']].drop_duplicates()
    return len(unique_bins)


def compute_seasonal_coverage_counts(df: pd.DataFrame) -> pd.Series:
    """Return counts by season for unique (year, season, lat, lon) bins."""
    df = df.copy()
    df['season'] = df['month'].apply(get_season)
    season_bins = df[['year', 'season', 'latitude', 'longitude']].drop_duplicates()
    return (
        season_bins['season']
        .value_counts()
        .reindex(['Summer', 'Autumn', 'Winter', 'Spring'])
        .fillna(0)
        .astype(int)
    )


# =========================
# Load & filter source data
# =========================
print("Loading data…")

# --- MATRIX ---
matrix = pd.read_csv(MATRIX_CSV, low_memory=False)

# Harmonise lat/lon names
matrix = matrix.rename(
    columns={"lat_center": "latitude", "lon_center": "longitude"}
)

# Derive year/month from month_center (monthly data)
matrix['month_center'] = pd.to_datetime(matrix['month_center'])
matrix['year'] = matrix['month_center'].dt.year
matrix['month'] = matrix['month_center'].dt.month

# Southern Ocean + time window
matrix = matrix[
    (matrix['latitude'] <= -30) &
    (matrix['year'] >= START_YEAR) &
    (matrix['year'] <= END_YEAR)
]

# Drop rows missing any required matrix feature
matrix_input_columns = [
    'latitude', 'longitude', 'month_center', 'temperature', 'salinity',
    'MLD', 'sla_median', 'rrs412', 'rrs443', 'rrs490', 'rrs555', 'rrs670',
    'PAR_mean', 'u10', 'v10', 'wind_speed', 'atm_co2', 'year', 'month',
    'doy', 'decimal_year', 'doy_radians', 'doy_sin', 'doy_cos', 'x_cart',
    'y_cart', 'z_cart', 'bottom_depth_m'
]
matrix = matrix.dropna(subset=matrix_input_columns)


# --- SOCAT (monthly, gridded) ---
socat_df = pd.read_csv(SOCAT_CSV, low_memory=False)
socat_df = socat_df.rename(
    columns={"lat_center": "latitude", "lon_center": "longitude"}
)

socat_df['month_center'] = pd.to_datetime(socat_df['month_center'])
socat_df['year'] = socat_df['month_center'].dt.year
socat_df['month'] = socat_df['month_center'].dt.month

socat_df = socat_df[
    (socat_df['latitude'] <= -30) &
    (socat_df['year'] >= START_YEAR) &
    (socat_df['year'] <= END_YEAR)
]

# Drop cells with no SOCAT pCO2 (adjust column name if different)
if 'pco2' in socat_df.columns:
    socat_df = socat_df.dropna(subset=['pco2'])
elif 'pCO2' in socat_df.columns:
    socat_df = socat_df.dropna(subset=['pCO2'])

# --- FLOATS (monthly, gridded) ---
floats_df = pd.read_csv(FLOATS_CSV, low_memory=False)
floats_df = floats_df.rename(
    columns={"lat_center": "latitude", "lon_center": "longitude"}
)

floats_df['month_center'] = pd.to_datetime(floats_df['month_center'])
floats_df['year'] = floats_df['month_center'].dt.year
floats_df['month'] = floats_df['month_center'].dt.month

floats_df = floats_df[
    (floats_df['latitude'] <= -30) &
    (floats_df['year'] >= START_YEAR) &
    (floats_df['year'] <= END_YEAR)
]

# Drop cells with no float pCO2 (adjust column name if different)
if 'pco2' in floats_df.columns:
    floats_df = floats_df.dropna(subset=['pco2'])
elif 'pCO2' in floats_df.columns:
    floats_df = floats_df.dropna(subset=['pCO2'])

# Drop rows missing any required floats feature
floats_input_columns = [
    'FLOATS_temperature_median', 'FLOATS_salinity_median', 'MLD', 'sla_median',
    'rrs412', 'rrs443', 'rrs490', 'rrs555', 'rrs670',
    'PAR_mean',
    'u10', 'v10', 'wind_speed',
    'atm_co2',
    'decimal_year', 'doy_sin', 'doy_cos',
    'x_cart', 'y_cart', 'z_cart',
    'bottom_depth_m'
]
floats_df = floats_df.dropna(subset=floats_input_columns)


# Combined (SOCAT + Floats) on the monthly grid
combined_df = pd.concat(
    [
        socat_df[['year', 'month', 'latitude', 'longitude']],
        floats_df[['year', 'month', 'latitude', 'longitude']]
    ],
    ignore_index=True
).drop_duplicates()


# =========================
# Ocean mask & denominators
# =========================
print("Building ocean mask (Natural Earth 1:50m)…")
ocean_cells = build_ocean_cell_mask(LAT_RANGE, LON_RANGE)
num_ocean_cells = len(ocean_cells)
print(f"Ocean cells in domain (−90…-30°, 1° grid): {num_ocean_cells}")

# Denominators (ocean-only)
max_possible = num_ocean_cells * NUM_MONTHS               # monthly bins over all years
total_possible_per_season = num_ocean_cells * NUM_YEARS   # one bin per (year, season, lat, lon)

if max_possible == 0 or total_possible_per_season == 0:
    raise RuntimeError(
        "Ocean mask produced zero cells — check dependencies or domain settings."
    )


# =========================
# Prepare ocean-only bins
# =========================
matrix_bins   = prep_bins_ocean_only(matrix,      ocean_cells)
socat_bins    = prep_bins_ocean_only(socat_df,    ocean_cells)
floats_bins   = prep_bins_ocean_only(floats_df,   ocean_cells)
combined_bins = prep_bins_ocean_only(combined_df, ocean_cells)


# =========================
# Totals (2003–2024)
# =========================
matrix_covered   = compute_total_coverage_count(matrix_bins)
socat_covered    = compute_total_coverage_count(socat_bins)
floats_covered   = compute_total_coverage_count(floats_bins)
combined_covered = compute_total_coverage_count(combined_bins)

print("\n=== Total Spatiotemporal Coverage (2003–2024, ocean-only) ===")
print(f"MLP model matrix:            {100 * matrix_covered   / max_possible:8.4f}%")
print(f"SOCATv2025 (monthly grid):   {100 * socat_covered    / max_possible:8.4f}%")
print(f"BGC-Argo floats (monthly):   {100 * floats_covered   / max_possible:8.4f}%")
print(f"SOCATv2025 + BGC-Argo:       {100 * combined_covered / max_possible:8.4f}%")


# =========================
# Seasonal coverage
# =========================
def print_seasonals(name: str, bins_df: pd.DataFrame):
    counts = compute_seasonal_coverage_counts(bins_df)
    print(f"\n{name} seasonal coverage (≥1 month in season, ocean-only):")
    for season in ['Summer', 'Autumn', 'Winter', 'Spring']:
        count = int(counts.get(season, 0))
        pct = 100.0 * count / total_possible_per_season
        print(f"  {season:<6}: {pct:8.4f}%")


print_seasonals("MLP model matrix",             matrix_bins)
print_seasonals("SOCATv2025 (monthly grid)",    socat_bins)
print_seasonals("BGC-Argo floats (monthly)",    floats_bins)
print_seasonals("SOCATv2025 + BGC-Argo floats", combined_bins)

print("\nDone.")
```
