### FCode for Figure 1

```python
import pandas as pd

# === Load SOCAT Data ===
socat_path = "/home/ldelaigue/Documents/Python/SOCAT-OSSE-SO/data/SOCATv2025_SO_clean.csv"
df_socat = pd.read_csv(socat_path, low_memory=False)

# === Load BGC-Argo Data ===
bgc_path = "/home/ldelaigue/Documents/Python/SOCAT-OSSE-SO/archive/MODEL_FLOATS/processing_steps/BGC-ARGO_mean_0-10m_surface_pco2_all_profiles_US_fco2.csv"
df_bgc = pd.read_csv(bgc_path, low_memory=False)
```

```python
import pandas as pd

# === Convert SOCAT date columns to datetime and filter ===
df_socat["date"] = pd.to_datetime(dict(
    year=df_socat["year"],
    month=df_socat["month"],
    day=df_socat["day"]
), errors="coerce")

df_socat = df_socat[df_socat["date"] <= "2022-12-31"].copy()

# === Convert BGC-Argo date columns to datetime and filter ===
df_bgc["date"] = pd.to_datetime(dict(
    year=df_bgc["year"],
    month=df_bgc["month"],
    day=df_bgc["day"]
), errors="coerce")

df_bgc = df_bgc[df_bgc["date"] <= "2022-12-31"].copy()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# =========================
# Prep
# =========================
df_bgc['datetime'] = pd.to_datetime(df_bgc['datetime'])
df_bgc['month'] = df_bgc['datetime'].dt.month
df_bgc['year']  = df_bgc['datetime'].dt.year


# Restrict latitude coverage
df_bgc  = df_bgc[df_bgc['latitude']  <= -35].copy()
df_socat = df_socat[df_socat['latitude'] <= -35].copy()

# Seasons
def get_season(m):
    if m in [12, 1, 2]:
        return 'Summer'
    elif m in [3, 4, 5]:
        return 'Autumn'
    elif m in [6, 7, 8]:
        return 'Winter'
    elif m in [9, 10, 11]:
        return 'Spring'

df_socat['season'] = df_socat['month'].apply(get_season)
df_bgc['season']   = df_bgc['month'].apply(get_season)

# Summer / Winter subsets for maps
summer_months = [12, 1, 2]
winter_months = [6, 7, 8]
df_socat_summer = df_socat[df_socat['month'].isin(summer_months)]
df_socat_winter = df_socat[df_socat['month'].isin(winter_months)]
df_bgc_summer   = df_bgc[df_bgc['month'].isin(summer_months)]
df_bgc_winter   = df_bgc[df_bgc['month'].isin(winter_months)]

# 1°x1° bins
for _df in (df_socat, df_bgc):
    _df['lat_bin'] = _df['latitude'].astype(int)
    _df['lon_bin'] = _df['longitude'].astype(int)

# Unique grid cells per year
socat_cells_per_year = (
    df_socat.drop_duplicates(subset=['year', 'lat_bin', 'lon_bin'])
    .groupby('year').size().sort_index()
)
bgc_cells_per_year = (
    df_bgc.drop_duplicates(subset=['year', 'lat_bin', 'lon_bin'])
    .groupby('year').size().sort_index()
)

# Year axis
all_years = list(range(1985, 2023))
socat_cells_per_year = socat_cells_per_year.reindex(all_years, fill_value=0)
bgc_cells_per_year   = bgc_cells_per_year.reindex(all_years,   fill_value=0)

# Seasonal hist helper
def seasonal_hist(df):
    df_unique = df.drop_duplicates(subset=['year', 'lat_bin', 'lon_bin'])
    seasonal_counts = df_unique.groupby(['year', 'season']).size().unstack(fill_value=0)
    for s in ['Summer', 'Autumn', 'Winter', 'Spring']:
        if s not in seasonal_counts.columns:
            seasonal_counts[s] = 0
    return seasonal_counts[['Summer', 'Autumn', 'Winter', 'Spring']].reindex(all_years, fill_value=0)

socat_seasonal = seasonal_hist(df_socat)
bgc_seasonal   = seasonal_hist(df_bgc)

season_colors = {
    'Summer': '#22223B',
    'Autumn': '#507DBC',
    'Winter': '#44FFD2',
    'Spring': '#DAF5FF'
}

# =========================
# Figure (5 panels: a–e)
# =========================
fig = plt.figure(figsize=(10, 14), dpi=300)

# GridSpec layout
gs = fig.add_gridspec(nrows=3, ncols=2, height_ratios=[1.2, 1.0, 1.0], hspace=0.28, wspace=0.18)

# --- (a) Summer map ---
ax_a = fig.add_subplot(gs[0, 0], projection=ccrs.SouthPolarStereo())
ax_a.set_extent([-180, 180, -90, -30], crs=ccrs.PlateCarree())
ax_a.coastlines(resolution='110m', linewidth=0.7)
ax_a.add_feature(cfeature.LAND, facecolor='lightgray', zorder=0)
ax_a.gridlines(draw_labels=False, linewidth=0.2, color='gray', alpha=0.5)

socat_a = ax_a.scatter(
    df_socat_summer['longitude'], df_socat_summer['latitude'],
    s=1, c='k', marker='.', alpha=0.8, transform=ccrs.PlateCarree(),
    label='SOCAT v2025', zorder=2
)
bgc_a = ax_a.scatter(
    df_bgc_summer['longitude'], df_bgc_summer['latitude'],
    s=1, c='tab:blue', marker='.', alpha=0.8, transform=ccrs.PlateCarree(),
    label='BGC-Argo floats', zorder=1
)
ax_a.text(0, 1.02, 'a)', transform=ax_a.transAxes, fontsize=14, va='bottom')

# --- (b) Winter map ---
ax_b = fig.add_subplot(gs[0, 1], projection=ccrs.SouthPolarStereo())
ax_b.set_extent([-180, 180, -90, -30], crs=ccrs.PlateCarree())
ax_b.coastlines(resolution='110m', linewidth=0.7)
ax_b.add_feature(cfeature.LAND, facecolor='lightgray', zorder=0)
ax_b.gridlines(draw_labels=False, linewidth=0.2, color='gray', alpha=0.5)

socat_b = ax_b.scatter(
    df_socat_winter['longitude'], df_socat_winter['latitude'],
    s=1, c='k', marker='.', alpha=0.8, transform=ccrs.PlateCarree(),
    label='SOCAT v2025', zorder=2
)
bgc_b = ax_b.scatter(
    df_bgc_winter['longitude'], df_bgc_winter['latitude'],
    s=3, c='tab:blue', marker='.', alpha=0.8, transform=ccrs.PlateCarree(),
    label='BGC-Argo floats', zorder=1
)
ax_b.text(0, 1.02, 'b)', transform=ax_b.transAxes, fontsize=14, va='bottom')

# Legend inside subplot (b)
fig.legend(
    handles=[socat_a, bgc_a],
    loc='upper center',
    bbox_to_anchor=(0.5, 0.63),  # tweak if needed
    ncol=2, frameon=True, fontsize=10, markerscale=7
)

# --- (c) Full-width bars: unique 1°x1° cells per year ---
ax_c = fig.add_subplot(gs[1, :])
width = 0.4
ax_c.bar([y - width/2 for y in all_years], socat_cells_per_year, width=width, label='SOCAT v2025', color='k')
ax_c.bar([y + width/2 for y in all_years], bgc_cells_per_year,   width=width, label='BGC-Argo floats', color='tab:blue')
ax_c.set_ylabel('Number of 1°×1° grid cells with ≥1 new $\\it{f}$CO$_2$ obs')
ax_c.grid(True, linestyle=':', alpha=0.5)
ax_c.text(0, 1.08, 'c)', transform=ax_c.transAxes, fontsize=14, va='top')
ax_c.set_ylim(0, 2500)
ax_c.legend()

# --- (d) SOCAT seasonal ---
ax_d = fig.add_subplot(gs[2, 0])
bottom = np.zeros(len(all_years))
for season in ['Summer', 'Autumn', 'Winter', 'Spring']:
    ax_d.bar(all_years, socat_seasonal[season], bottom=bottom,
             label=season, color=season_colors[season], width=0.8)
    bottom += socat_seasonal[season].values
ax_d.set_ylabel('Number of 1°×1° grid cells with ≥1 new $\\it{f}$CO$_2$ obs')
ax_d.set_xlabel('Year')
ax_d.grid(True, linestyle=':', alpha=0.5)
ax_d.text(0, 1.08, 'd)', transform=ax_d.transAxes, fontsize=14, va='top')
ax_d.set_ylim(0, 2500)
ax_d.legend(loc='upper left')

# --- (e) BGC-Argo seasonal ---
ax_e = fig.add_subplot(gs[2, 1])
bottom = np.zeros(len(all_years))
for season in ['Summer', 'Autumn', 'Winter', 'Spring']:
    ax_e.bar(all_years, bgc_seasonal[season], bottom=bottom,
             label=season, color=season_colors[season], width=0.8)
    bottom += bgc_seasonal[season].values
ax_e.set_xlabel('Year')
ax_e.grid(True, linestyle=':', alpha=0.5)
ax_e.text(0, 1.08, 'e)', transform=ax_e.transAxes, fontsize=14, va='top')
ax_e.set_ylim(0, 2500)
# no legend to avoid duplication

# --- Final layout ---
plt.tight_layout()
plt.savefig("/home/ldelaigue/Documents/Python/SOCAT-OSSE-SO/figs/SO_fCO2_5panel_maps_plus_coverage.png", dpi=300, bbox_inches='tight')
plt.show()
```
