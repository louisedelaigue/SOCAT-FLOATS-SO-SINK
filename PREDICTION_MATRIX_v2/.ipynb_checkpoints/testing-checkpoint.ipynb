```python
import pandas as pd

output_path = (
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/"
    "PREDICTION_MATRIX_v2/processing_steps/"
    "ARMOR3D_TSMLD_SouthernOcean_2003_2024_monthly_1deg_feature_engineered_monthly_1deg.csv"
)
df = pd.read_csv(output_path)
```

```python
df.columns
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Make sure month_center is datetime
df["month_center"] = pd.to_datetime(df["month_center"])

# Keep only February 2023
df_feb2023 = df[
    (df["month_center"].dt.year == 2023) &
    (df["month_center"].dt.month == 2)
].copy()

# Convert to 2D grid: rows=lat, cols=lon
sal_grid = df_feb2023.pivot(
    index="lat_center",
    columns="lon_center",
    values="MLD"
)

# Sort axes so the map displays correctly
sal_grid = sal_grid.sort_index().sort_index(axis=1)

# Extract coordinates
lats = sal_grid.index.values
lons = sal_grid.columns.values
sal = sal_grid.values

# Plot
fig = plt.figure(figsize=(14, 7))
ax = plt.axes(projection=ccrs.Robinson())
ax.set_global()

pcm = ax.pcolormesh(
    lons,
    lats,
    sal,
    transform=ccrs.PlateCarree(),
    shading="auto",
    vmin=20,
    vmax=300
)

ax.coastlines(linewidth=0.8)
ax.add_feature(cfeature.LAND, facecolor="lightgray")
ax.gridlines(draw_labels=False, linewidth=0.5, linestyle="--", alpha=0.5)

cbar = plt.colorbar(pcm, ax=ax, orientation="horizontal", pad=0.05, shrink=0.8)
cbar.set_label("Sea surface salinity")

ax.set_title("Salinity from df - February 2023")
plt.show()
```

```python
import xarray as xr

armor = (
    "/remote/unity/bgc-output/DELAIGUE/SOCAT-FLOATS-OSSE/data/ARMOR-3D/"
    "cmems_obs-mob_glo_phy_my_0.125deg_P1M-m_so-to-mlotst_179.94W-179.94E_82.19S-20.06S_0.00m_2003-01-01-2024-12-01.nc"
)

ds = xr.open_dataset(armor)
```

```python
ds
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Select February 2023 surface temperature
so = ds["mlotst"].sel(time="2023-02-01")

# Create figure and map
fig = plt.figure(figsize=(14, 7))
ax = plt.axes(projection=ccrs.Robinson())
ax.set_global()

# Plot
pcm = ax.pcolormesh(
    ds["longitude"],
    ds["latitude"],
    so,
    transform=ccrs.PlateCarree(),
    shading="auto"
)

# Map features
ax.coastlines(linewidth=0.8)
ax.add_feature(cfeature.LAND, facecolor="lightgray")
ax.add_feature(cfeature.BORDERS, linewidth=0.3)
ax.gridlines(draw_labels=False, linewidth=0.5, linestyle="--", alpha=0.5)

# Colorbar
cbar = plt.colorbar(pcm, ax=ax, orientation="horizontal", pad=0.05, shrink=0.8)

# Title

plt.show()
```
