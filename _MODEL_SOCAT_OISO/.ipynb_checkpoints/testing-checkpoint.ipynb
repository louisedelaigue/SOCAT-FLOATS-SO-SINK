```python
import pandas as pd
import re

# Path to the file in the Jupyter environment
file_path = "/home/ldelaigue/Documents/Python/SOCAT-OSSE-SO/MODEL_SOCAT_OISO/data/Expocodes-OISO-dans-SOCAT.xlsx"

# Load the Excel file into a DataFrame
df = pd.read_excel(file_path)

# Southern Hemisphere season classifier
def periode_to_sh_season(p):
    p = p.lower()

    # Summer (DJF)
    if any(m in p for m in ["dec", "jan", "fev", "feb"]):
        return "summer"

    # Autumn (MAM)
    if any(m in p for m in ["mars", "mar", "avr", "apr", "mai", "may"]):
        return "autumn"

    # Winter (JJA)
    if any(m in p for m in ["jun", "jui", "juil", "jul", "aout", "août", "aug"]):
        return "winter"

    # Spring (SON)
    if any(m in p for m in ["sep", "sept", "oct", "nov"]):
        return "spring"

    return None  # fallback

# Apply to dataframe
df["season"] = df["période"].apply(periode_to_sh_season)

# Extract the cruise number from the Campagne column
df["cruise_number"] = df["Campagne"].apply(
    lambda x: int(re.findall(r"OISO[-]?(\d+)", x)[0])
)

df.rename(columns={"Unnamed: 3": "start_month", "Expocode SOCAT":"expocode"}, inplace=True)
```
